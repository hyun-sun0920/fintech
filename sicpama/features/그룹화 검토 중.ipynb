# 자세하게 보기

## ① test 그룹

```python
test_brief = test.drop(['dateOfBirth', 'isMarketingAgreed','first_visit','last_visit'], axis=1)
```

```python
test_brief.info()
```

```python
for id in test_brief['storeId'].unique():
    for seg in test_brief['segment'].unique():
        print('='*10, f'id : {id}, seg : {seg}','='*10)
        display(test_brief[(test_brief['storeId']==id) & (test_brief['segment']==seg)])
```

## ② control 그룹

```python
control_brief = control.drop(['dateOfBirth', 'isMarketingAgreed','first_visit','last_visit'], axis=1)
```

```python
control_brief.info()
```

```python
for id in control_brief['storeId'].unique():
    for seg in control_brief['segment'].unique():
        print('='*10, f'id : {id}, seg : {seg}','='*10)
        display(test_brief[(test_brief['storeId']==id) & (test_brief['segment']==seg)])
```
