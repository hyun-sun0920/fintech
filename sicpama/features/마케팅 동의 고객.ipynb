# Feature Engineering
* 첫방문으로부터 얼마나 지났는지
* 마지막 방문으로부터 얼마나 지났는지
* Y : 최근 90일이내에 방문했는지

```python
today = pd.Timestamp('2025-05-14')
```

```python
# 문자열이라면 datetime으로 변환
data['first_visit'] = pd.to_datetime(data['first_visit'])
data['last_visit']  = pd.to_datetime(data['last_visit'])

# 경과 일수 계산
data['days_since_first'] = (today - data['first_visit']).dt.days
data['days_since_last']  = (today - data['last_visit']).dt.days
```

```python
data.head()
```

## Y label (최근 90일이내 재방문)

```python
data['revisited'] = (pd.to_datetime(data['last_visit']) >= today - pd.Timedelta(days=90)).astype(int)
```

```python
data.to_csv('../data/feature통합.csv', index=False)
```

```python
data[data['revisited']==1]
```

```python
numeric_df = data.loc[:,['visit_count', 'avg_revisit_days', 'pagerank', 'degree','betweenness', 'closeness', 'eigen', 'paidAmount', 'days_since_first','days_since_last','revisited']]
```

```python
numeric_df.head()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# corr: pandas.DataFrame, numeric_df.corr() 로 구한 상관계수 행렬
plt.figure(figsize=(10,8))
sns.heatmap(numeric_df.corr(), 
            annot=True,        # 셀 안에 값 표시
            fmt=".2f",         # 소수점 둘째 자리 포맷
            cmap="coolwarm",   # 컬러맵
            linewidths=0.5,    # 셀 경계 두께
            square=True)       # 정사각형 셀
plt.title("Features Correlation Heatmap")
plt.tight_layout()
plt.show()
```
