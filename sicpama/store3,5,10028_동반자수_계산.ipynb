# store=3의 동반자 수(동일 인물 여러 번 초대해도 1번으로 계산)

```python
from neo4j import GraphDatabase
import pandas as pd
from IPython.display import display

# Neo4j 연결
driver = GraphDatabase.driver("bolt://localhost:7687", auth=("neo4j", "gustjs21@"))

def run_query(tx, query):
    return tx.run(query).data()

# 방향 없는 동반자 수 쿼리 (store 3 기준)
query = """
MATCH (c:Customer)-[:`DINED_WITH_3`]-(d:Customer)
WHERE id(c) < id(d)
UNWIND [c, d] AS person
RETURN COALESCE(person.id, toString(ID(person))) AS customerId,
       COUNT(*) AS companion_count
"""

with driver.session(database="sicpama") as session:
    result = session.execute_read(run_query, query)

# DataFrame으로 변환
df_companion = pd.DataFrame(result)

# 출력
print("✅ store 3 기준, 고객별 동반자 수:")
display(df_companion)
```
