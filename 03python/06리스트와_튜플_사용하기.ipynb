# 튜플 tuple
* 튜플은 리스트처럼 모든 자료형을 넣어 만들 수 있다.
* 순서가 있음 -> 인덱싱, 슬라이싱 가능
* 자료의 추가, 삭제가 안됨 -> 그래서 리스트로 바꾼 뒤, 다시 튜플로
* ()안에 자료를 넣어서 만든다.
* 튜플 형변환 함수는 tuple()
* 자료 1개만 넣어서 만들 수 있다. (1,) 형식임
* 튜플 생성 시 자료가 2개 이상인 경우 ()는 생략 가능

```python
t=(1,2,3)
t
```

```python
t[0]
```

```python
t[::-1]
```

```python
t2=(4,5,6)
```

# 시퀀스 객체(자료형)
* 문자열, 리스트, 튜플 같은 순서가 있는 자료형: 시퀀스 자료형
* 순서가 있고, 인덱스 번호가 부여되며 인덱싱, 슬라이싱이 가능
* 반복문에 시퀀스 자료형을 그대로 넣어서 반복시키는 것이 가능

```python
t2
```

```python
* 특정 값이 있는지 확인 in, not in
```

```python
4 in t2
```

```python
7 in t2
```

```python
c = "시퀀스 자료형은 in, not in을 자료가 내부에 있는지 확인 가능하다"
```

```python
"자료" in c
```

```python
list_01=[1,2,3,4,5]
3 in list_01
```

* 시퀀스 객체는 +연산자를 이용해서 서로 연결이 가능
