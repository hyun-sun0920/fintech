# 반복문 for, while
* 반복문은 동일한 작업을 여러 번 반복해야 할 경우 사용
* for, while 두 종류의 반복문이 있음
* for는 반복횟수가 정해져 있는 경우 사용
* while은 반복 횟수가 정해져 있지 않은 경우(무한 반복)사용: 특정 조건 만날 때 멈춤

## for 반복문: 반복할 횟수가 정해져 있는 경우
for 변수1, 변수2 in 반복할 객체/자료(문자열, 리스트, 튜플, 딕셔너리:<br>
    반복할 코드

```python
# 숫자 생성 함수 range(시작, 끝, step)
```

```python
#for문을 이용해서 숫자 1-10을 출력
for num in range(1,11):
    print(num, end=" ")
```

* 순서가 있는 자료형(문자열, 리스트, 튜플) + 순서는 없지만 dict도 반복문에서 사용가능

```python
# 문자열
for char in "파이썬은 문자열을 반복문에 넣어 출력 가능":
    print(char, end="")
```

```python
#리스트
for l_item in [1,2,3,"사","오",[6,7,8],(8,9,10),{"key":"value"}]:
    print(l_item)
```

```python
# 딕셔너리
d = dict(이름="홍길동", 나이=40, 주소="서울", 전화번호="010-1111-1111")
```

* 1부터 50까지 수 중에서 짝수인 수의 합을 구하시오.
* 1-50 수를 만들기 (range(1-51))
* 홀수, 짝수 구분 if %2==0
* 짝수 일때 누적합, 결과 저장할 변수 hap
* 최종 결과 출력

```python
hap=0
for num in range(1,51):
    if num%2==0:
        print(f"hap:{hap},num:{num}")
        hap=hap+num
print(hap)
```

## continue, break
* continue: 코드 실행을 건너뛰고 반복 계속 진행
* break: 특정 조건이 되었을 때 정지

```python
for idx, i in enumerate(s):
    if idx%2==0:
        continue
    else:
        print(i)
```

```python
num=0
i=1
while 1<=i<=50:
    print(i,end=" ")
    if i%3==0 or i%2==0:
        print(f"hap: {hap}, i:{i}")
        num=num+i
    i=i+1
print(num)    
```

# While의 무한 반복
* while 뒤의 조건이 True인 경우 무한 반복
* while문 내에서 if 조건문으로 break를 걸어주어야 멈춤

```python
while True:
    print("무한반복", end=" ")
```

* 무한 반복 후 while문 내에서 조건을 줘서 멈추기

```python
i=1
while True:
    print("조건 반복", end=" ")
    if i>=100:
        break
    i+=1
```

* 무한 반복이 되는 조건:True
* True가 되는 경우: 1, 정수, 실수, 문자열, list, tuple, dict, True인 경우
* False가 되는 경우: 0, 0.0, ""(공백 아님), '', [],(),{}, False

```python
i=1
while 0:
    print(i, "조건 반복")
    if i>=100:
        break
    i+=1
```

# 중첩 반복문
* 반복문을 2개 이상 겹쳐서 사용
* 바늘 시계의 원리와 같음
* 안쪽 반복문이 끝난 후 바깥쪽 반복문이 실행
* 초침 -> 분침 -> 시침

* 구구단 만들기 2중 중첩 for문
* 2단- 9단
* 1-9
* 2*1=2

```python
for num1 in range(2,10):
    for num2 in range(1,10):
        print(f"{num1}x{num2}=", num1*num2)
```

* 반복문 3개로 시계 만들기
* 초 00-59
* 분 00-59
* 시 00-23

```python
import time
for num1 in range(0,24):
    for num2 in range(0,60):
        for num3 in range(0,60):
            print(f"{num1:02d}시 {num2:02d}분 {num3:02d}초")
            time.sleep(0.22)
```
