# 리스트 표현식
* 리스트 안에 for 반복문과 if else 조건문을 한 번에 적어서 리스트를 만드는 방법
* 여러 줄의 코드를 1줄로 줄일 수 있다.
* for문과 if else 문을 사용해서 처리할 때보다 속도가 빠르다.
* [실행할 명령 for 변수 in list, tuple]

```python
for i in range(1,11):
    print("짝수입니다.") if i%2==0 else print("홀수입니다")
```

* 1부터 10까지 들어있는 리스트 만들기

```python
a1=[]
for i in range(1,11):
    a1.append(i)
a1
```

```python
[i for i in range(1,11)]
```

* 1에서 10까지의 숫자 중 짝수만 넣어서 리스트 만들기

# 파일 읽기 쓰기 open(), with open()
## 1)파일에 문자열을 쓰거나 읽을 때는 open 함수 사용
* 파일저장
변수명=open("파일이름", 모드(w:쓰기, r:읽기)<br>
변수명.write("저장할 내용")<br>
변수명.close()<br>
close()를 안하면 계속 열려있음

```python
file=open("test.txt", "w")
file.write("파일 만들기 연습")
file.close()
```

변수1.close()

## with open
* with open은 close()를 하지 않아도 자동으로 close()처리를 해 줌

```python
with open("test2.txt",'w') as file3:
    file3.write("with open을 사용하면 파일을 자동으로 닫아주어 편리하다.")
```

```python
with open("test2.txt",'r') as file4:
    data=file4.read()
print(data)
```

# encoding에 따른 읽기 오류
* 운영체제나(windows, linux, macOS), 한글/영어 윈도우에 따라서 인코딩 되는 포멧이 달라짐
* 인코딩에 따른 오류가 발생 가능
* web표준은 utf-8이므로 인코딩은 utf-8로 하는 것이 좋다.
* windows의 경우 인코딩이 cp949이므로 주의

```python
with open("test_utf-8.txt",'w',encoding='utf-8') as file5:
    file5.write("이것은 with open으로 만든 utf-8입니다.")
```

```python
with open('test_utf-8.txt','r', encoding='utf-8') as file6:
    file6_data=file6.read()
print(file6_data)
```

# json 파일 읽어와서 내용 정리하기

```python
with open("./data/서울특별시_관광지입장정보_2011_2016.json", 'r',encoding="utf-8") as file7:
    trip_info=file7.read()
trip_info
```

```python
print(type(trip_info))
```

```python
import json
```

```python
data=json.loads(trip_info)
```

```python
len(data)
```

```python
for item in data[0]:
    print(item)
```

```python
for item in data[0].items():
    print(item)
```

```python
for item in data[:2]:
    for key in item:
        print(key)
```

```python
for item in data[:2]:
    for key in item:
        print(item[key])
```

```python
* value를 리스트로 만든 후 dict함수로 딕셔너리 만들어 데이터프레임 만들기
```

```python
year_mon=[]
sido=[]
place=[]
foriners=[]
total=[]

for item in data[:2]:
    foriners.append(item['ForNum'])
    total.append(item['NatNum'])
    place.append(item['resNm'])
    sido.append(item['sido'])
    year_mon.append(item['yyyymm'])
    
tour_info = dict(년월=year_mon, 시도=sido, 장소=place, 외국인=foriners, 전체방문객수= total)
tour_info
```

```python
import pandas as pd
```

```python
df=pd.DataFrame(tour_info)
df
```

* 딕셔너리로 바로 만들어서 데이터프레임으로 변환

* setdefault 함수를 이용해 만들기

```python
dic_result={}
for dicts in data:
    for key, value in dicts.items():
    dic_result.setdefault(key, [].append(value))
dic_result
```
