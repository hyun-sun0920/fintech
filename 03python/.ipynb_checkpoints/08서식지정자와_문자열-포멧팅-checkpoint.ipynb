# 서식지정자와 format 메서드
* 문자열로만 입력해야 하는 상황에서 같은 위치에 문자를 여러번 바꿔야 할 때 사용
* 문자의 자릿수 맞춤, 위치 정렬에도 사용

# 1. 서식지정자 %d, %s, %f, %길이s

### 1)문자를 입력할 때 %s

```python
print("이 수업의 이름은 %s 입니다." % "국어")
```

```python
for i in ["국어","수학","사회","영어"]:
    print("이 수업의 이름은 %s 입니다." %i)
```

### 2) 정수를 입력할 때 %d decimal(10진수, 정수)

```python
print("올해는 %d년 입니다." %2025)
```

### 3)실수를 입력할 때 %f 기본 소숫점 6자리로 표시

```python
print("사람의 정상 체온은 약 %f도 입니다" %36.5)
```

* 소숫점 표기를 조절할 때는 %소숫점자리수f ex)%2f

```python
print("사람의 정상 체온은 약 %.2f도 입니다" %36.5)
```

### 4) 서식지정자로 문자열 길이, 정렬 %길이s

```python
print("%10s" % "10자리의 자리수 확보의 확보의 확보")
```

```python
print(150000)
print(15000)
```

```python
print("가격은 %10d 입니다." %150000)
print("가격은 %10d 입니다." %15000)
```

### 5)여러 항목 동시에 넣기

```python
print("오늘은 %4d년 %2d월 %2d일 입니다." %(2025, 3, 17))
```

```python
a, b, c= (2025, 3, 17)
print(a,b,c)
```

```python
company=['삼성전자','LG전자','현대차','대한항공','하나투어']
stock=[71800,121000,195000,31900,86600]
```

```python
for com, st in zip(company,stock):
    print("%s의 주가는 %d 입니다." %(com, st))
```

# 2. format 메서드 사용하기, f-string 문자열 포메팅

### 1) format 메서드 사용하기
* "문자열{인덱스}문자열 끝".format(값)

```python
print("{0}의 주가는 {1} 입니다.".format('삼성전자',71800))
```

```python
print("{1}의 주가는 {0}입니다.").format('삼성전자',71800)
```

```python
print("{0}의 주가는 {1}원입니다.\n{0}의 상한가는 {2}\n{0}의 하한가는 {3}입니다.".format("삼성전자",71500,50200,93000))
```

### f-string으로 값 넣기 python 3.6 사용가능

```python
company
```

```python
stock
```

```python
print(f"{company[0]}의 주가는 {stock[0]} 입니다.")
```

```python
sam="삼성전자"
sto=71800
print(f"{sam}의 주가는 {sto}입니다.")
```

```python
for com, sto in zip(company, stock):
    print("{0}의 주가는 {1}원입니다".format(com, sto))
```

```python
for com, sto in zip(company, stock):
    print(f"{com}의 주가는 {sto}입니다.")
```

# 3. format 메서드와 f-string으로 문자열 정렬하기, 자리수 지정해 출력하기

* "{인덱스:<길이}".format(값)
* f"{값:>길이}"

<공간을 10칸 확보하고 글자를 왼쪽 정렬

```python
print("{0:<10}".format("삼성전자"))
```

```python
print("{0:>10}".format("삼성전자"))
```

```python
"%10s" %'삼성전자'
```

```python
f"{'삼성전자':<10}"
```

```python
f"{'삼성전자':>10}"
```

* 숫자 표기시 숫자 자리수 맞추고 기본 값 지정하기
* "{인덱스:채울숫자 길이 서식지정자}".format(값), 단, 서식 지정자에서 %는 빼고 적는다.
* f"{값:채울 숫자 길이 서식지정}"%는 뺌

```python
print("{0}월".format(1))
```

```python
print("{0:03d}월".format(12))
```

```python
print(f"{10:03d}월")
```

* 실수의 정수자리수와 소수점 자리수 지정하기
* ("{인덱스: 기본값 자리수 서식지정자}".format(값)
* f"{값:기본값 자리수 서식지정자}"
* 실수에서 자리수는 정수 +.+소숫점


```python
print("{0:03.3f}".format(2.245678))
```

```python
print("{0:07.3f}".format(2.245678))
```

```python
print("{0:07.6f}".format(2.245678))
```

```python
import time
for si in range(0,12): 
   # print(f"{si:02d}시")
    for bun in range(0, 60):
        print(f"{si:02d}시 {bun:02d}분")
        time.sleep(0.2)
```

* 공백을 원하는 문자로 채우고 자리수 맞추기
* {인덱스:원하는 숫자/문자 > <자리수}.format(값)
* f"{값:원하는 숫자/문자 > <자리수}"
