#파이썬의 자료형(data type)
##파이썬의 자료형은 동적으로 변화
* int: 정수(소수점이 없는 숫자)
* float: 실수(소수점이 있는 숫자)
* str: 문자/문자열
* list: 여러 개의 자료가 순서대로 저장된 자료형. 모든 자료형을 넣어 만들 수 있음
* tuple: list와 비슷하나 한 번 만들면 수정이 안됨. 모든 자료형을 넣을 수 있음
* dictionery(dict): key: value 쌍으로 이루어진 자료형. 모든 자료형을 넣을 수 있음

```python
a=3
b="3"
```

```python
a
```

```python
print(a,b)
```

* 정수로 형변환 int()

```python
a + int(b)
```

* 실수로 변환 float()

```python
a+float(b)
```

* 문자열로 변환 str()

```python
str(a)+b
```

*리스트 형변환 list()

```python
t=(1,2,3,4)
print(t, type(t))
```

```python
t2=list(t)
print(t2,type(t2))
```

```python
t2.append(5)
print(t2)
```

* 튜플로 형변환 

```python
t3=tuple(t2)
print(t3,type(t3))
```

# input(): 키보드 입력을 받는 함수
* input("화면에 표시될 메시지")
* 입력 받는 값은 공백으로 구분해서 여러개를 받을 수 있음
* 입력 받는 값은 모두 str 문자열로 받음

```python
input_name=input("이름을 입력해 주세요")
print(input_name)
```

```python
input_name2 = input("이름을 입력해 주세요")
print(input_name2)
```

```python
data = input("숫자를 3개 입력해주세요")
print(data, type(data))
```

* input().split()으로 여러 데이터를 list 형태로 받기

```python
data = input("숫자를 3개 입력해주세요").split()
print(data, type(data))
```

```python
data[0]+data[1]
```

```python
a, b, c, d, e = input("숫자를 5개 입력해주세요").split()
print(a,b,c,d, type(a),type(b),type(c),type(d))
```
