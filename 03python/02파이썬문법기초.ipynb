# print문의 옵션
* sep: 구분자-print 문 안에 여러개의 자료를 출력할 때 자료를 구분해주는 문자
* 기본값은 "" 공백

```python
print("이것은", '파이썬', '입니다.')
```

```python
print("이것은", '파이썬', '입니다.', sep="_")
```

```python
print("헐크","아이어맨","스파이더맨",sep="vs")
```

*end옵션 - print가 끝나고 실행될 옵션
*기본값=\n-이스케이프 문자 new line 새줄

```python
print("헐크","아이어맨","스파이더맨",sep="vs")
print("헐크","아이어맨","스파이더맨",sep="vs")
```

```python
#end값을 공백""으로 준 경우
print("헐크","아이어맨","스파이더맨",sep="vs", end="   ")
print("헐크","아이어맨","스파이더맨",sep="vs")
```

```python
#end값을 공백"\t" (tap)으로 준 경우
print("헐크","아이어맨","스파이더맨",sep="vs", end="\t")
print("헐크","아이어맨","스파이더맨",sep="vs")
```

```python
#end값을 공백 "\r" (carriage return)으로 준 경우: 밑에 줄이 위에 줄 덮음
print("헐크","아이어맨","스파이더맨",sep="vs", end="\r")
print("헐크","아이어맨","스파이더맨",sep="vs")
```

```python
#end값을 공백"\n" (new line)으로 준 경우
print("헐크","아이어맨","스파이더맨",sep="vs", end="\n\n")
print("헐크","아이어맨","스파이더맨",sep="vs")
```

#변수
* 자료를 담는 택배상자
* 택배 상자는 1개만 넣을 수 있다.
* 변수는 영문과 숫자를 조합해서 만들 수 있다.
* 변수명의 시작은 반드시 영문자로 시작 or _로 시작
* 변수명이 숫자로 시작하거나 -등의 특수문자로 시작하면 오류
* 예약어(이미 파이썬에서 사용하고 있는 함수명, 문법용 단어) ex) for, list, if, id는 변수명으로 사용할 수 없다.

*주의!! 예악어를 써도 파이썬에서는 변수가 만들어진다. -> id가 변수가 되어버리는 상황발생
*그러므로 예악어를 잘 기억해두자!!

*변수명은 스네이크 표기법으로 만든다.
*변수명=자료
*=는 같다라는 표시가 X, 등호 오른쪽의 자료를 왼쪽 변수에 할당한다.
