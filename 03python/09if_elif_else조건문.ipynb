# if else 조건문

## 1. 조건문이란? 특정 조건을 판별해서 작업을 분기시키는 문법
if 조건식:<br>
    조건이 참일 때 실행할 코드
<br>else: <br>
    조건이 거짓일 때 실행할 코드

```python
x=13
```

```python
if x==10:
    print(f"{x}는, 10입니다.")
else:
    print(f"{x}는, 10이 아닙니다.")
```

```python
l1=['파이썬',"조건문",10,15.3,True,[1,2,3],(4,5,6),{"key": 365}]
```

```python
if "파이썬" in l1:
    print("l1안에 파이썬이 있습니다.")
else:
    print("l1안에 파이썬이 없습니다.")
```

* 숫자를 입력받아 홀수인지 짝수인지 판단하는 프로그램

```python
x=input("숫자를 입력하세요")
if int(x)%2==0:
    print(f"입력한 숫자 {x}는 짝수입니다.")
else:
    print(f"입력한 숫자 {x}는 홀수입니다.")
```

```python
x=input()
if 0 <= int(x) <=10 and int(x)%2==0:
    print(f"{x}는 0에서 10사이에 있는 숫자이고 짝수입니다.")
else:
    print(f"{x}는 0에서 10사이에 있는 숫자가 아닙니다.")
```

## 2. 다중 조건문 사용하기 elif
if 조건식:
    조건이 참일 때 실행할 코드
elif 조건식2:
    조건식2가 참일 때 실행할 코드
elif 조건식 3:
    조건식3이 참일 때 실행할 코드
else:
    조건식 1,2,3이 모두 거짓일 때 실행할 코드
    

* 점수를 입력받아서 81-100 A, 61-80 B, 41-60 C, 21-40 D, 0-20 F 등급

```python
score=int(input("점수를 입력하세요"))
if 0<=score <=100:
    if 0<=score <=20:
        print(f"{score}는 F등급 입니다.")
    elif 21 <= score <=40:
        print(f"{score}는 D등급 입니다.")
    elif 41<= score <= 60:
        print(f"{score}는 C등급 입니다.")
    elif 61<= score <= 80:
        print(f"{score}는 B등급 입니다.")
    else:
        print(f"{score}는 A등급 입니다.")
else:
    print("숫자를 0~100사이로 입력해 주세요")
```

* 1-50사이의 숫자를 입력받아서 3의 배수면 3의 배수입니다. 3의 배수가 아니면 3의 배수가 아닙니다.
    * 입력받기, 정수로 변환하기
    * 1-50 사이의 숫자인가?
    * 3의 배수인가?

```python
a=int(input("숫자입력"))
if 1<=a<=50:
    if a%3==0:
        print(f"{a}는 3의 배수입니다.")
    else:
        print(f"{a}는 3의 배수가 아닙니다.")
else:
    print("1~50 사이의 숫자가 아닙니다.")
```

## 3. 조건 표현식
* 참일 때, 실행코드 if 조건 else 거짓일 때 실행코드

```python
*일반적인 if else 문
num = int(input("숫자를 입력하세요"))
print(num,type(num))
if num%2==0:
    print("짝수")
else:
    print("홀수")
```

```python
# 조건 표현식
num = int(input("숫자를 입력하세요"))
print(num,type(num))
print("짝수") if num%2==0 else print("홀수")
```
