```python
from haversine import haversine, Unit
```

```python
def calc_distance(row):
    home=(row['lat'], row['long'])
    store = (row['merch_lat'], row['merch_long'])
    return haversine(home, store, unit=Unit.KILOMETERS)

data['distance'] = data.apply(calc_distance, axis=1)
```

# 홀드아웃

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.4, stratify=y, random_state=10)
```

# 모델 생성

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import classification_report
```

```python
for i in range(1, 11): 
    dtc = DecisionTreeClassifier(max_depth=i, random_state=10)
    print(i)
    dtc.fit(X_train, y_train)
    pred = dtc.predict(X_test)
    print(classification_report(y_test, pred))
```
