* 나이를 구간화해서 생존율과의 관계 파악
    * 10, 20, 30, 40, 50
    * 0 -10 under10s
    * 10-19 10s
    * 20-29 20s

```python
def age_cat(x):
    if x < 10:
        return "under10s"
    elif 10 <=x<20:
        return "10s"
    elif 20 <=x<30:
        return "20s"
    elif 30 <=x<40:
        return "30s"
    elif 40 <=x<50:
        return "40s"
    elif 50 <=x<60:
        return "50s"
    elif 60 <=x<70:
        return "60s"
    elif 70 <=x<80:
        return "80s"
    elif 80 <=x<90:
        return "80s"
```

```python
data['age_cat']=data['Age'].apply(age_cat)
data.head()
```

# 상관관계 분석 .corr()
* 변수들 간의 상관관계 분석
* 1이면 강한 양의 상관관계
* -1이면 강한 음의 상관관계
* 0이면 관계가 없음
* 0.5나 -0.5면 고묀
* 0.7이상이면 두 변수 중 한개만 선택
* method='pearson': 숫자형 데이터들 간의 상관관계 분석
* method='spearman': 범주형 데이터가 섞여 있는 데이터셋의 경우 스피어만으로 상관관계 분석

```python
df2.corr(method='spearman')
```
