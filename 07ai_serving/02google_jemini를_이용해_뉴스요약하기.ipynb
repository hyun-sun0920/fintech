```python
import gradio as gr
import os
from google import genai
from dotenv import load_dotenv

# 환경변수에서 api 키 불러오기
load_dotenv()
api_key=os.getenv("GEMINI_API_KEY")


# Gemini 응답 함수
def gemini_chat(user_message):
    response = client.models.generate_content(
        model="gemini-2.0-flash",
        contents=user_message
    )
    return response.text

# Gradio UI 구성
with gr.Blocks() as demo:
    gr.Markdown("## 💬 Gemini 2.0 Flash Chatbot")

    with gr.Row():
        with gr.Column(scale=4):
            input_text = gr.Textbox(label="메시지를 입력하세요", placeholder="예: 인공지능의 미래는?")
            submit_btn = gr.Button("Gemini에게 질문하기")
        with gr.Column(scale=6):
            output_text = gr.Textbox(label="Gemini의 응답", lines=10)

    # 버튼 클릭 시 Gemini 호출
    submit_btn.click(fn=gemini_chat, inputs=input_text, outputs=output_text)
    input_text.submit(fn=gemini_chat, inputs=input_text, outputs=output_text)

demo.launch()
```
