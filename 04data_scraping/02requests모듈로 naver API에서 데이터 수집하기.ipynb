# 3. requests 모듈 사용법

* 1) 모듈 불러오기: import requests
* 2) url 변수에 저장
* 3) url 파라미터
    * payload={파라미터1: value, 파라미터2: value}
* 4) header
    * headers={key:value}
* 5) HTTP 요청 보내기 get, post
    * get: r=requests.get(url, params=payload, headers=headers)
    * post: r=requests.post(url, data={key:value}): 로그인 해서 크롤링 하는 경우
* 6) 만들어진 url 확인: print(r.url)
* 7) 응답 코드 확인: print(r.status_code)
* 8) 응답 요소 출력: r.text, r.json(), r.content

```python
def json_unpack(data):
    result={}
    for item in data['items']:
        for key, value in item.items():
            result.setdefault(key, []).append(value)
    df=pd.DataFrame(result)
    return df
```

```python
keyword=input("검색하고 싶은 책이름을 입력하세요")
url = "https://openapi.naver.com/v1/search/book.json"
headers={"X-Naver-Client-Id" : "HaXEGE3cbTofjrDFePfE", "X-Naver-Client-Secret": "U1uuXe2QfQ"}
payload={'query':keyword,'display': 100, 'start': '1', 'sort':'sim'}
r=requests.get(url, params=payload, headers=headers)
data=r.json()
print("전체 검색 데이터 수: ", data['total'], '검색할 페이지수: ', (int(data['total'])/100)+1)
result_df=json_unpack(data)
result_df
```

# 블로그, 뉴스, 책 데이터 각각 저장하기

```python
import time
keyword=input("검색할 키워드를 입력해주세요")
response_blog=[]
response_news=[]
response_book=[]
for category in ['blog','news','book']:
    page=1
    start_num=1
    while True:
        url=f"https://openapi.naver.com/v1/search/{category}.json"
        Client_Id="HaXEGE3cbTofjrDFePfE"
        Client_Secret="U1uuXe2QfQ"
        payload=dict(query=keyword, display=100, start=start_num, sort='sim')
        headers={"X-Naver-Client-Id": Client_Id , "X-Naver-Client-Secret": Client_Secret}
        r=requests.get(url, params=payload, headers=headers)
        print(r.url, end='\r')
        print(r.status_code, end='\r')
        response=r.json()
        total_pages=(response['total']//100)+1
        
        if category=='blog':
            response_blog.append(response)
        elif category=='news':
            response_news.append(response)
        elif category=='book':
            response_book.append(response)
            
        result_json.append(response)
        if start_num==1000: 
            break
        if start_num<901:
            start_num+=100
        elif start_num==901:
            start_num+=99

        if page<total_pages:
            page+=1
        else:
            break
        time.sleep(0.5)
print(len(response_blog))
print(len(response_news))
print(len(response_book))
```

```python
import pandas as pd
pd.DataFrame(response_blog)
```

```python
def json_unpack(all_data):
    final_result={}
    for page in all_data:
        for item in page['items']:
            for key, value in item.items():
                final_result.setdefault(key, []).append(value)
    df=pd.DataFrame(final_result)
    return df
```

```python
for jsons, cat in zip([response_blog, response_news, response_book],["blog", "news", "book"]):
    unpack_result=json_unpack(jsons)
    unpack_result.to_csv(f"naver_{cat}_result.csv", encoding="utf-8", index=False)
```
