```python
# 네이버 검색 API 예제 - 블로그 검색
# 네이버 검색 API 예제 - 블로그 검색
import os
import sys
import urllib.request
client_id = "HaXEGE3cbTofjrDFePfE"
client_secret = "U1uuXe2QfQ"
encText = urllib.parse.quote("핀테크")
url = "https://openapi.naver.com/v1/search/blog?query=" + encText + "&display=30"# JSON 결과.보통은 10개 나오지만 여기서는 30개
# url = "https://openapi.naver.com/v1/search/blog.xml?query=" + encText # XML 결과
request = urllib.request.Request(url)
request.add_header("X-Naver-Client-Id",client_id)
request.add_header("X-Naver-Client-Secret",client_secret) 
response = urllib.request.urlopen(request) # 응답코드가 옴
rescode = response.getcode()
if(rescode==200): #오류번호가 200번대이면
    response_body = response.read()
    print(response_body.decode('utf-8')) #디코딩해서 우리가 알아볼 수 있는 문자로 바꿔라
else:
    print("Error Code:" + rescode)
```

```python
def text_clean(x):
    result=x.replace("<b>","").replace("</b>","")
    return result
```

```python
result={}
for item in data['items']:
    for key, value in item.items():
        value=text_clean(value)
        result.setdefault(key,[]).append(value)
result
```

```python
df=pd.DataFrame(result)
df
```
